```python
import nltk
nltk.download('stopwords')
from nltk.corpus import stopwords
import re
```

LOWERCASING TEXT

```python
sentence = "His dog's name is Spark"
lower_cased = sentence.lower()
print(lower_cased)
```

```python
sentence_list = ["Could you bring my water?",
                 "Where's she going?",
                 "That's not fair"
                ]

lower_cased_sentence = [x.lower() for x in sentence_list]
print(lower_cased_sentence)
```

REMOVING STOP WORDS

```python
en_stopwords = stopwords.words('english')
print(en_stopwords)
```

```python
sentence = "it was too far to go to the shop and he did not want her to walk"
sentence_no_stopwords = " ".join([word for word in sentence.split() if word not in en_stopwords])
print(sentence_no_stopwords)
```

```python
en_stopwords.remove("did")
en_stopwords.remove("too")
en_stopwords.append("want")

sentence_list_nosw = [word for word in sentence.split() if word not in en_stopwords]
print(sentence_list_nosw)
```

REGEX

```python
result_search = re.search("pattern", r"string to contain the pattern")
print(result_search)
```

```python
result_search = re.search("pattern", r"string to contain the past tense")
print(result_search)
```

```python
string = r"sara was able to help me find the items I needed quickly"
new_string = re.sub("sara", "sarah", string)
print(new_string)
```

```python
customer_reviews = ["david was a nice customer to sarah",
                    "daniel is a naughty",
                    "I don't think I like sara",
                    "where is she",
                    "who is sarah?",
                    "ambigous is she"
                    ]

string_s = r"sarah?"

reviews_for_sarah = []

for string in customer_reviews:
    if re.search(string_s, string):
        reviews_for_sarah.append(string)

print(reviews_for_sarah)
```

```python
a_reviews = []

string_a = r"^a"

for string in customer_reviews:
    if(re.search(string_a, string)):
        a_reviews.append(string)

print(a_reviews)
```

```python
y_reviews = []
string_y = r"y$"

for string in customer_reviews:
    if(re.search(string_y, string)):
        y_reviews.append(string)

print(y_reviews)
```

```python
whowhere_reviews = []
string_wh = r"(who|where)"

for string in customer_reviews:
    if(re.search(string_wh, string)):
        whowhere_reviews.append(string)

print(whowhere_reviews)
```

```python
no_punctuation_reviews = []

no_punct = r"[^\w\s]"

for string in customer_reviews:
    new_no_punct = re.sub(no_punct, "", string)
    no_punctuation_reviews.append(new_no_punct)

print(no_punctuation_reviews)
```

STANDARDIZING TEXT - STEMMING

```python
from nltk.stem import PorterStemmer
```

```python
ps = PorterStemmer()

connect_list = ['connected', 'connectivity', 'connecting', 'connection']

for word in connect_list:
    print(word, ":", ps.stem(word))
```

```python
learn_list = ['learn', 'learner', 'learns', 'learned', 'learning']

for l in learn_list:
    print(l, ":", ps.stem(l))
```

STANDARDIZING TEXT - LEMMATIZATION

```python
nltk.download('wordnet')
from nltk.stem import WordNetLemmatizer
```

```python
lemmatizer = WordNetLemmatizer()

for word in connect_list:
    print(word, ":", lemmatizer.lemmatize(word))
```

```python
for word in learn_list:
    print(word, ":", lemmatizer.lemmatize(word))
```

N-GRAMS

```python
import nltk, pandas as pd, matplotlib as plt
```

```python
sentence_ngram = "the rise of artificial intelligence has led to advancements\
    in natural language processing, computer vision, and other fields.\
    machine learning algorithms are becoming more sophisticated, enabling computers to perform complex tasks, fields. fields."

sentence_ngram_tokens = sentence_ngram.split()

print(sentence_ngram_tokens)
```

```python
unigrams = (pd.Series(nltk.ngrams(sentence_ngram_tokens, 1)).value_counts())
print(unigrams.sort_values())
```

```python
unigrams[:10].sort_values().plot.barh()
```

```python
bigrams = (pd.Series(nltk.ngrams(sentence_ngram_tokens, 2)).value_counts())
print(bigrams)
```

```python
trigrams = (pd.Series(nltk.ngrams(sentence_ngram_tokens, 3)).value_counts())
print(trigrams)
```
