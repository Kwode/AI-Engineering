LOGISTIC REGRESSION

```python
import pandas as pd 
from sklearn.feature_extraction.text import CountVectorizer
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score, classification_report
```

```python
sentiment_data = pd.read_csv('sentiment_dataset.csv')
sentiment_data.head(20)
data = sentiment_data.sample(frac=1).reset_index(drop=True)
data.head(5)
```

```python
x = data['text']
y = data['label']

count_vec = CountVectorizer()
count_fit = count_vec.fit_transform(x)
count_array = count_fit.toarray()
bow = pd.DataFrame(count_array, columns=count_vec.get_feature_names_out())
bow
```

SPLIT DATA TO TRAINING AND TESTING SETS

```python
x_train, x_test, y_train, y_test = train_test_split(bow, y, test_size=0.3, random_state=42)
```

```python
lr = LogisticRegression(random_state=7).fit(x_train, y_train)
```

```python
y_pred = lr.predict(x_test)
accuracy_score(y_pred, y_test)
print(classification_report(y_pred, y_test, zero_division=0))
```

NAIVE BAYES

```python
from sklearn.naive_bayes import MultinomialNB
```

```python
nb = MultinomialNB()
nb.fit(x_train, y_train)
y_pred_nb = nb.predict(x_test)
accuracy_score(y_pred_nb, y_test)
```

LINEAR SUPPORT VECTOR MACHINES

```python
from sklearn.linear_model import SGDClassifier
```

```python
svm = SGDClassifier()
svm.fit(x_train, y_train)
svm_y_pred = svm.predict(x_test)
accuracy_score(svm_y_pred, y_test)
```
