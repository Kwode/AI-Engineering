```python
import nltk
from nltk.tokenize import word_tokenize
from nltk.stem import PorterStemmer, WordNetLemmatizer
from nltk.corpus import stopwords
import re
import pandas as pd
```

```python
dataset = pd.read_csv("tripadvisor-hotel-reviews.csv")
dataset['lowercased_review'] = dataset['Review'].str.lower()
dataset.head()
# dataset['Review'][0]
```

Remove Stopwords

```python
en_stopwords = stopwords.words('english')

dataset['no_stopwords'] = dataset['lowercased_review'].apply(lambda x: " ".join([word for word in x.split() if word not in en_stopwords]))

dataset.head()
```

Removing Punctuations

```python
dataset['reviews_no_punct'] = dataset.apply(lambda x: re.sub(r"[*]", " star", x['no_stopwords']), axis=1)
dataset['reviews_no_punct'] = dataset.apply(lambda x: re.sub(r"[^\w\s]", "", x['reviews_no_punct']), axis=1)
dataset.head()
```

Tokenization

```python
dataset['tokenized_review'] = dataset.apply(lambda x: word_tokenize(x['reviews_no_punct']), axis=1)
dataset.head()
```

Stemming & Lemmatization

```python
ps = PorterStemmer()
dataset['stemmed_review'] = dataset['tokenized_review'].apply(lambda tokens: [ps.stem(token) for token in tokens])
```

```python
lem = WordNetLemmatizer()
dataset['lemmatized_review'] = dataset['tokenized_review'].apply(lambda tokens: [lem.lemmatize(token) for token in tokens])
dataset.head()
```

N-grams

```python
lematized_list = sum(dataset['lemmatized_review'], [])

unigrams = pd.Series(nltk.ngrams(lematized_list, 1)).value_counts()
print(unigrams.sort_values())
```

```python
bigrams = pd.Series(nltk.ngrams(lematized_list, 2)).value_counts()
print(bigrams)
```
