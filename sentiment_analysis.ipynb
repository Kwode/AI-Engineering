Rule-Based Sentiment Analysis

```python
sentence_1 = 'I had a great time at the movie. It was really funny'
sentence_2 = 'I had a great time at the movie but the parking was terrible'
sentence_3 = "I had a great time at the movie but the parking was wasn't great"
sentence_4 = "I want to see a movie"
```

TEXTBLOB

```python
from textblob import TextBlob
```

```python
sentiment_sentence_1 = TextBlob(sentence_1)
print(sentiment_sentence_1.sentiment.polarity)
```

```python
sentiment_sentence_2 = TextBlob(sentence_2)
print(sentiment_sentence_2.sentiment.polarity)
```

```python
sentiment_sentence_3 = TextBlob(sentence_3)
print(sentiment_sentence_3.sentiment.polarity)
```

```python
sentiment_sentence_4 = TextBlob(sentence_4)
print(sentiment_sentence_4.sentiment.polarity)
```

VADER

```python
from vaderSentiment.vaderSentiment import SentimentIntensityAnalyzer
vaderSentiment = SentimentIntensityAnalyzer()
```

```python
print(vaderSentiment.polarity_scores(sentence_1))
print(vaderSentiment.polarity_scores(sentence_2))
print(vaderSentiment.polarity_scores(sentence_3))
print(vaderSentiment.polarity_scores(sentence_4))
```

Pre-Trained Transformer

```python
import transformers
from transformers import pipeline
```

```python
sentiment_pipeline = pipeline('sentiment-analysis')
```

```python
sentiment_pipeline(sentence_1)
sentiment_pipeline(sentence_2)
sentiment_pipeline(sentence_3)
sentiment_pipeline(sentence_4)
```

```python
specific_model = pipeline('sentiment-analysis', model='finiteautomata/bertweet-base-sentiment-analysis')
```

```python
print(sentence_4)
specific_model(sentence_4)
```
